<h3> Rotina Principal </h3>

Tokenização por frases (Sentence tokenization)

```python
from nltk.tokenize import sent_tokenize
```

```python
# Forma 1 - Usando NTLK
texto = "Olá. Gostaria de falar algo. Eu aprecio a Inteligência Artificial. Eu desejo aprender muito com o professor Roger. Espero ter sucesso na jornada?"
frases = sent_tokenize(texto, language='portuguese')
print(f"Qtde de tokens: {len(frases)}")
i = 1
for frase in frases:
  print(f"TÓKEN {i}: {frase}")
  i+=1
```

Tokenização por palavras (Word-Level)

```python
from nltk.tokenize import word_tokenize

texto = "O Mercadinho do Seu Zé usa Inteligência Artificial para analisar feedbacks!"

# Tokenização
tokens_nltk = word_tokenize(texto, language='portuguese')

print(f"Quantidade de tókens: {len(tokens_nltk)}")
print(tokens_nltk)
```

Tokenização por sub-palavras

```python
!pip install tokenizers
```

```python
from tokenizers import Tokenizer
from tokenizers.models import BPE
from tokenizers.pre_tokenizers import Whitespace
from tokenizers.trainers import BpeTrainer

# 1. Instancia um modelo BPE em branco
tokenizer = Tokenizer(BPE(unk_token="[UNK]"))

# Define como o texto é pré-dividido (por espaços em branco)
tokenizer.pre_tokenizer = Whitespace()

# 2. Configura o treinador do BPE
trainer = BpeTrainer(special_tokens=["[UNK]"], vocab_size=50)

# 3. Treina o tokenizador usando o seu arquivo que está na pasta raiz do Colab
caminho_arquivo = "corpus_treino.txt"  # Substitua pelo nome do seu arquivo .txt
tokenizer.train(files=[caminho_arquivo], trainer=trainer)

# 4. Executa a tokenização por subpalavras
palavra = "recontextualização"
output = tokenizer.encode(palavra)

print(f"Palavra original: {palavra}")
print(f"Subpalavras obtidas: {output.tokens}")
```
