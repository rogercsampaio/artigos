<h5> Adicionando tabelas </h5>

```python
# Lendo tabelas
print(f"\033[94mℹ️\033[0m Informação: Qtde de tabelas {len(doc.tables)}")
if(len(doc.tables) > 0):
    for tabela in doc.tables:
        print(tabela)
else:
    print(f"\033[94mℹ️\033[0m Informação: Sem tabelas encontradas")
```

```python
# Adicionando uma tabela
# Especificando o número e linhas e colunas
num_linhas = 3
num_colunas = 4

# Criando a estrutura da tabela
tabela = doc.add_table(rows=num_linhas, cols=num_colunas)

# Populando a tabela. 
for i in range(num_linhas):
    for j in range(num_colunas):
        # Acessa cada célula da tabela e define o texto
        tabela.cell(i, j).text = f"Linha {i+1}, Coluna {j+1}"
```

```python
def imprimir_tabela(tabela):
    for linha in tabela.rows:
        # Para cada linha, percorre as células e imprime o texto delas
        linha_impressao = [celula.text.strip() for celula in linha.cells]
        print(" | ".join(linha_impressao))
imprimir_tabela(tabela)
```

```python
# Criando outra tabela, porém de outra maneira: 1. dataframe, 2. conversão para tabela
import pandas as pd
machine_learning_desc = "Machine Learning é uma área da inteligência artificial que permite que computadores aprendam padrões a partir de dados e façam previsões ou decisões sem serem programados explicitamente para cada tarefa."
deep_learning_desc = "Deep Learning é um subconjunto de Machine Learning que usa redes neurais artificiais com muitas camadas para aprender representações complexas dos dados, sendo ideal para tarefas como reconhecimento de imagens, voz e linguagem natural."

dados_dict = [['machine learning',machine_learning_desc], ['deep learning',deep_learning_desc]]
conceitos_ia_df = pd.DataFrame(data = dados_dict,columns = ['termo','significado'])
print(f"Tipo: {type(conceitos_ia_df)}")
conceitos_ia_df
```

```python
# Adicionar tabela (com cabeçalho + dados)
tabela2 = doc.add_table(rows=1 + len(conceitos_ia_df), cols=len(conceitos_ia_df.columns))

# Preencher cabeçalho
for i, col_name in enumerate(conceitos_ia_df.columns):
    tabela2.cell(0, i).text = col_name

# Preencher linhas
for row_idx, row in enumerate(conceitos_ia_df.values, start=1):
    for col_idx, value in enumerate(row):
        tabela2.cell(row_idx, col_idx).text = str(value)
```

```python
imprimir_tabela(tabela2)
```

```python
doc.save("bases/o_que_ia_2.docx")
print(f"Tabela adicionada")
```
